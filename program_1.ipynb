```python
import pandas as pd

data = {
    "Name":[ "Asha" , "Ravi" , "Meera" , "Karen" ],
    "Marks": [88,72,91,65],
    "Branch": [ "CSE" , "ECE" , "CSE" , "ISE" ]
}
```

```python
df = pd.DataFrame(data)

print("\n --- Pandas DataFrame---")
print(df)
```

```python
print("\n--- Cloumn Indexing ---")

print("Name column:")
print(df["Name"])
print("\nMarks Column")
print(df["Marks"])
```

```python
print("\n --- Row Indexing ---")
print("First Row:")
print(df.iloc[0])
print("\n Third Row:")
print(df.iloc[2])
```

```python
print("\n--- Specific Value----")
print("Marks odf First Student:",df.iloc[0,1])
```

```python
print("\n---- Row Slicing ---")
print("First Two Rows:")
print(df.iloc[0:2])
```

```python
print("\n --- Row and Column Slicing---")
print("First Three Rows with Names and Marks:")
print(df.iloc[0:3,[0,1]])
```

```python
print("\n--- Data Filtering ---")
result = df[df["Marks"] >70]
print("Students with Marks Greater then 70:")
print(result)
```
